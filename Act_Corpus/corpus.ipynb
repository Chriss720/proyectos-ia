# Corpus y vocabulario compartidos:

### Corpus minimo

```python
# 1. Definición del Corpus: 
# Un conjunto de programas en texto plano usado para entrenar o construir el vocabulario
CORPUS = r'''
def factorial(n):
    """Calcula el factorial de un número n."""
    if n <= 1:
        return 1
    return n * factorial(n - 1)
def fibonacci(n):
    """Genera el n-ésimo número de la secuencia de Fibonacci."""
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a
def es_primo(num):
    """Verifica si un número entero es primo."""
    if num <= 1:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True
def es_palindromo(cadena):
    """Verifica si una cadena de texto es un palíndromo."""
    cadena = cadena.lower().replace(" ", "")
    return cadena == cadena[::-1]
class Punto:
    """Clase que representa un punto en un plano cuadrático 2D."""
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def distancia_origen(self):
        """Calcula la distancia de este punto al origen (0,0)."""
        return (self.x ** 2 + self.y ** 2) ** 0.5
for i in range(10):
    print(i, fibonacci(i))
'''
# 2. Creación del Vocabulario:
# Obtenemos todos los caracteres únicos que existen en nuestro CORPUS
chars = sorted(set(CORPUS))
# 'stoi' (string to integer): Diccionario para convertir cada caracter a un número entero único
stoi = {ch: i for i, ch in enumerate(chars)}
# 'itos' (integer to string): Diccionario para convertir de número entero de vuelta a su caracter
itos = {i: ch for ch, i in stoi.items()}
# Tamaño total de nuestro vocabulario (cantidad de caracteres únicos)
VOCAB_SIZE = len(chars)
# 3. Funciones de Codificación y Decodificación:
def encode(s: str):
    """
    Convierte una cadena de texto a una lista de números enteros 
    usando el diccionario stoi (tokenización a nivel de caracteres).
    """
    return [stoi[c] for c in s]
def decode(ids):
    """
    Convierte una lista de números enteros de regreso a una cadena de texto 
    usando el diccionario itos.
    """
    return "".join(itos[i] for i in ids)
# Imprimir métricas básicas del corpus
print("VOCAB_SIZE:", VOCAB_SIZE, "| len(CORPUS):", len(CORPUS))
```

### Ventanas (entrada / objetivo del siguiente caracter) 

```python
import numpy as np

SEQ = np.array(encode(CORPUS), dtype=np.int64)
block_size = 32
X_rows, Y_rows = [], []
for i in range(0, len(SEQ) - block_size):
    X_rows.append(SEQ[i : i + block_size])
    Y_rows.append(SEQ[i + 1 : i + 1 + block_size])

X_np = np.stack(X_rows)   # (N, T)
Y_np = np.stack(Y_rows)   # (N, T) — en t, objetivo = siguiente carácter tras X_np[n,t]
print("X_np:", X_np.shape, "Y_np:", Y_np.shape)
```
